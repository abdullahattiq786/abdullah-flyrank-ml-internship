*Classification, clustering, ranking, or scoring — which one, and why?*


**Task type: Ranking**

I am framing Lane 2 — Refresh / Content Opportunity Scoring as a ranking problem.

The goal is to rank pages by how strongly the available evidence suggests that they should be reviewed first for refresh, expansion, protection, pruning, or monitoring.

This is a ranking problem rather than a simple classification problem because the practical output is an ordered review queue. ## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*


For the starter ML task, I will use the current downward-trend label as a proxy target:

`is_declining_label = trend_direction == "down"`

This means that a page is labeled positive when its observed `trend_direction` is `"down"`.

This is a **proxy**, not a true future outcome. It describes what is currently observed in the available window rather than proving that the page will decline in the future.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*


A fixed rule such as "review every page whose trend is down" is easy to understand, but it ignores other useful signals.

A page can have different combinations of visibility, impressions, sessions, age, freshness, CTR, position, engagement, and trend. For example, a declining page with substantial search exposure may deserve more attention than a low-volume page with the same trend label.

ML can learn combinations of multiple observable signals and produce a ranked ordering rather than relying on one manually chosen condition.

I would still start with a transparent rule-based baseline. ML is useful only if it can improve the ranking quality enough to justify the additional complexity.

The model should remain decision-support: it should help a reviewer decide which pages to investigate first, not claim that an edit will definitely cause recovery.
