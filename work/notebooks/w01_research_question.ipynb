# Week 1 — Research Question and Lane Selection

## Chosen Lane
Lane 2 — Refresh / Content Opportunity Scoring


## 1. Search Question

**Which pages should be reviewed first for refresh, expansion, protection, pruning, or monitoring?**

The goal is to determine whether observable content and search-performance signals can be used to prioritize pages for human review.

The system is intended to support prioritization rather than guarantee that a particular SEO action will improve a page's performance.

## 2. Unit of Analysis

**One row represents one content/page observation.**

The decision will be made at the page/content level. Each page can be evaluated using observable signals such as impressions, clicks, sessions, CTR, average position, content age, freshness, engagement, and trend information.

The purpose is to rank individual pages according to how strongly the available evidence suggests that they deserve review.


## 3. Output

The intended output is a **ranked review queue of pages**.

For each page, the eventual output should include:

- a priority/opportunity score;
- a suggested action such as refresh, expansion, protection, pruning, or monitoring;
- reason codes explaining why the page received its score;
- a confidence label.

The output should help a human reviewer understand both **which pages to review first** and **why they were prioritized**.

## 4. Action

A FlyRank reviewer could use the ranked queue to decide which pages to investigate first.

Depending on the signals associated with a page, the reviewer could consider actions such as refreshing content, expanding content, protecting a strong page, pruning a weak page, or monitoring the page.

The model or scoring system would support human decision-making rather than automatically applying an SEO change.

## 5. Cost of a Wrong Recommendation

A false positive could cause a reviewer to spend limited time investigating a page that does not represent a meaningful opportunity. This could delay work on pages that deserve more attention.

A false negative could also be costly because a page with a genuine opportunity or risk might be ranked too low and receive less attention.

Because reviewer capacity is limited, the quality of the highest-ranked recommendations is especially important.

Therefore, evaluation should focus on ranking quality, including metrics such as Precision@K and average precision, rather than relying only on overall classification accuracy.


## 6. Why Data / ML Can Help

A content inventory can contain a large number of pages, making it difficult for a human reviewer to compare every page consistently.

The starter dataset contains observable signals about content and search performance. These signals can be combined to identify pages that may deserve review.

A transparent rule-based score can provide a simple baseline. Machine learning can then be used to test whether combinations of multiple signals improve the ranking of useful review candidates.

The purpose of using data and ML is therefore not to replace human judgment, but to reduce the search space and direct limited reviewer attention toward pages that appear most worth investigating.

```python
num_pages = len(df)

print(f"Number of page observations: {num_pages:,}")
```

```python
high_visibility_pages = (df["impressions_90d"] >= 500).sum()

print(
    f"Pages with at least 500 impressions in the last 90 days: "
    f"{high_visibility_pages:,}"
)
```

```python
declining_pages = (df["trend_direction"] == "down").sum()

print(
    f"Pages currently marked with a downward trend: "
    f"{declining_pages:,}"
)
```

```python
high_visibility_pct = high_visibility_pages / len(df) * 100
declining_pct = declining_pages / len(df) * 100

print(f"Pages with >=500 impressions: {high_visibility_pct:.2f}%")
print(f"Pages with downward trend: {declining_pct:.2f}%")
```
